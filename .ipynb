```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python

# Convert pickup column into datetime format
df['pickup'] = pd.to_datetime(df['pickup'])

# Sort data by pickup time
df = df.sort_values(by='pickup')

# Create line chart
plt.figure(figsize=(10,5))
plt.plot(df['pickup'], df['fare'])

plt.title('Fare Over Time')
plt.xlabel('Pickup Timestamp')
plt.ylabel('Fare')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
# Group data by pickup_borough and calculate total fare
borough_fare = df.groupby('pickup_borough')['fare'].sum()

# Create bar chart
plt.figure(figsize=(8,5))
borough_fare.plot(kind='bar')

plt.title('Total Fare by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.xticks(rotation=45)
plt.show()
```

```python
# Count trips for each payment method
payment_count = df['payment'].value_counts()

# Create pie chart
plt.figure(figsize=(7,7))

plt.pie(
    payment_count,
    labels=payment_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Trips by Payment Method')
plt.show()
```

```python
# Create histogram
plt.figure(figsize=(8,5))

plt.hist(df['distance'], bins=20, edgecolor='black')

plt.title('Distribution of Trip Distance')
plt.xlabel('Distance')
plt.ylabel('Frequency')

plt.grid(axis='y')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

df.boxplot(column='tip', by='pickup_borough')

plt.title('Tip Distribution by Pickup Borough')
plt.suptitle('')
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')

plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='pickup_borough')

plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='distance',
    y='fare',
    hue='pickup_borough'
)

plt.title('Distance vs Fare by Pickup Borough')
plt.xlabel('Distance')
plt.ylabel('Fare')

plt.show()
```

```python
# Select numerical columns
columns = ['distance', 'fare', 'tip', 'tolls', 'total']

# Create correlation matrix
corr = df[columns].corr()

# Create heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(
    df,
    vars=['distance', 'fare', 'tip', 'total'],
    hue='pickup_zone'
)

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x='payment',
    y='fare'
)

plt.title('Fare Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare Amount')

plt.show()
```
